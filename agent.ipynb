```python
from langchain.agents import create_agent
from langchain_groq import ChatGroq
COMPANY_NAME = "Tecproducts"
llm=ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0,
    max_retries=2,
    api_key="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"
)
def get_product(product:str)->str:
    """Get product details and pricing"""
    products = {
        # ADD YOUR PRODUCTS HERE (UNLIMITED!)
        "laptop": "Laptop Pro - ₹50,000, 16GB RAM, 1TB SSD",
        "phone": "Phone X - ₹25,000, 5G, 128GB",
        "watch": "Smart Watch - ₹5,000, GPS, Heart Monitor",
        # ... KEEP ADDING AS MANY AS YOU WANT!
    }
    return products.get(product.lower(), "Product not found")
def get_offer()->str:
    """Get current special deals"""
    return"50% off + free setup"
def convince_buyer(query:str)->str:
    """Generate convincing sales pitch"""
    return f"{query} is our bestseller from {COMPANY_NAME}. 3x ROI guaranteed!"

agent=create_agent(
    model=llm,
    tools=[get_product,get_offer,convince_buyer],
    system_prompt=f"you are persuasive sales agent for {COMPANY_NAME} who sells products and convinces customers"
)
```

```python
response=agent.invoke(
    {"messages":[{"role":"user","content":"tell me about your best product and discount"}]}
)
print(response['messages'][-1])
```
