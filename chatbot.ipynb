```python
from langchain.agents import create_agent
from langchain_groq import ChatGroq
llm=ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0,
    max_retries=2,
    api_key="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"
)
def get_weather(city:str)->str:
    """Get weather for a given city"""
    return f"its currently sunny in{city}!"
def get_my_name()->str:
    """Get the AI agent name."""
    return"Niki"
agent=create_agent(
    model=llm,
    tools=[get_weather,get_my_name],
    system_prompt="you are helpfull ssistant"
)
```
